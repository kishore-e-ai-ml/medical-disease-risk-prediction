# 01. Exploratory Data Analysis (EDA)
## Disease Prediction from Medical Data
This notebook demonstrates exploratory data analysis across three medical datasets: Heart Disease, Diabetes, and Breast Cancer.

```python
import os
import sys
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_dataset, get_dataset_summary
from src.config import DATASETS_CONFIG

%matplotlib inline
sns.set_theme(style='whitegrid')
```

### 1. Heart Disease Dataset Exploration

```python
df_heart = load_dataset('heart')
print('Heart Disease Data Shape:', df_heart.shape)
display(df_heart.head())
print('\nMissing values:\n', df_heart.isnull().sum())
print('\nTarget distribution:\n', df_heart['target'].value_counts())
```

### 2. Diabetes Dataset Exploration

```python
df_diabetes = load_dataset('diabetes')
print('Diabetes Data Shape:', df_diabetes.shape)
display(df_diabetes.head())
print('\nMissing values (zeros in physiological measures):\n', (df_diabetes[['Glucose', 'BloodPressure', 'BMI']] == 0).sum())
print('\nTarget distribution:\n', df_diabetes['Outcome'].value_counts())
```

### 3. Breast Cancer Dataset Exploration

```python
df_bc = load_dataset('breast_cancer')
print('Breast Cancer Data Shape:', df_bc.shape)
display(df_bc.head())
print('\nTarget distribution (1=Malignant, 0=Benign):\n', df_bc['target'].value_counts())
```
